```python
# =========================================================
# PREPROCESSING DENGAN CACHE (SAVE STEMMING)
# UNTUK DIPAKAI BERULANG KALI TANPA ULANG DARI AWAL
# =========================================================

import pandas as pd
import re
import nltk
import os
import pickle
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from tqdm import tqdm

# =========================================================
# DOWNLOAD NLTK DATA
# =========================================================
nltk.download('punkt_tab', quiet=True)
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)

# =========================================================
# KONFIGURASI
# =========================================================
PROGRESS_DIR = '/content/drive/MyDrive/Penelitian_@/progress_testing'
os.makedirs(PROGRESS_DIR, exist_ok=True)

# File cache
CACHE_FILE = os.path.join(PROGRESS_DIR, 'stemming_cache.pkl')
CHECKPOINT_FILE = os.path.join(PROGRESS_DIR, 'checkpoint_data.csv')

# =========================================================
# FUNGSI PEMBERSIHAN TEKS
# =========================================================
def cleanse_text(text):
    if pd.isna(text):
        return ''
    text = str(text)
    text = re.sub(r'\[[^\]]+', '', text)
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'[\^\\s]', '', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'\s+', ' ', text)
    text = text.strip().lower()
    return text

# =========================================================
# NORMALISASI TOKEN
# =========================================================
def normalize_text(tokens):
    normal_words = {
        "isn't": "is not", "doesn't": "does not", "don't": "do not",
        "it's": "it is", "s": "is", "m": "am", "re": "are",
        "ve": "have", "ll": "will", "d": "would", "can't": "can not",
        "won't": "will not", "shouldn't": "should not"
    }
    return [normal_words.get(token, token) for token in tokens]

# =========================================================
# STOPWORDS
# =========================================================
stopwords_list = stopwords.words('english')

def remove_stopwords(tokens):
    return [word for word in tokens if word not in stopwords_list]

# =========================================================
# STEMMING DENGAN CACHE
# =========================================================
# Inisialisasi Sastrawi (hanya sekali)
factory = StemmerFactory()
stemmer = factory.create_stemmer()

# Load cache jika ada
if os.path.exists(CACHE_FILE):
    with open(CACHE_FILE, 'rb') as f:
        stem_cache = pickle.load(f)
    print(f"✓ Loaded {len(stem_cache)} kata dari cache")
else:
    stem_cache = {}
    print("✓ Cache baru dibuat")

def stem_with_cache(token_list):
    """Stemming dengan cache - CEPAT untuk proses ulang"""
    result = []
    for token in token_list:
        if token in stem_cache:
            result.append(stem_cache[token])
        else:
            stemmed = stemmer.stem(token)
            stem_cache[token] = stemmed
            result.append(stemmed)
    return result

# =========================================================
# LOAD DATASET
# =========================================================
print("\n" + "="*60)
print("📂 LOADING DATASET")
print("="*60)

df = pd.read_excel('/content/drive/MyDrive/Penelitian_@/data/data_twitter_rupiah_melemah.xlsx')

if 'text' not in df.columns:
    raise KeyError("Kolom 'text' tidak ditemukan dalam DataFrame.")

df['full_text'] = df['text'].astype(str)
print(f"✓ Dataset loaded: {len(df):,} rows")

# =========================================================
# RESUME DARI CHECKPOINT (JIKA PERNAH TERPUTUS)
# =========================================================
start_idx = 0
if os.path.exists(CHECKPOINT_FILE):
    checkpoint_df = pd.read_csv(CHECKPOINT_FILE)
    start_idx = len(checkpoint_df)
    print(f"✓ Resume dari checkpoint index ke-{start_idx}")
    # Load hasil yang sudah diproses
    df_processed = checkpoint_df.to_dict('records')
else:
    df_processed = []
    print("✓ Mulai preprocessing baru")

# =========================================================
# PROSES PIPELINE DENGAN PROGRESS BAR
# =========================================================
print("\n" + "="*60)
print("⚙️  PROSES PREPROCESSING (DENGAN CACHE)")
print("="*60)

for idx in tqdm(range(start_idx, len(df)), desc="Processing"):
    raw_text = df.iloc[idx]['full_text']

    # Pipeline
    clean = cleanse_text(raw_text)
    tokens = word_tokenize(clean)
    normalized = normalize_text(tokens)
    filtered = remove_stopwords(normalized)
    stemmed = stem_with_cache(filtered)

    hasil = ' '.join(stemmed)

    df_processed.append({
        'raw_text': raw_text,
        'hasil_preprocessing': hasil,
        'token_count': len(stemmed)
    })

    # Simpan checkpoint setiap 500 data (agar tidak rugi jika terputus)
    if (idx + 1) % 500 == 0:
        temp_df = pd.DataFrame(df_processed)
        temp_df.to_csv(CHECKPOINT_FILE, index=False)

        # Simpan cache juga
        with open(CACHE_FILE, 'wb') as f:
            pickle.dump(stem_cache, f)
        print(f"\n  ✓ Checkpoint saved at {idx+1} rows | Cache: {len(stem_cache)} kata")

# =========================================================
# FINAL PROCESSING
# =========================================================
print("\n" + "="*60)
print("📊 FINAL PROCESSING")
print("="*60)

final_df = pd.DataFrame(df_processed)

# Hapus data kosong
before_empty = len(final_df)
final_df = final_df[final_df['hasil_preprocessing'].str.strip() != '']
after_empty = len(final_df)

# Hapus duplikat
before_dup = len(final_df)
final_df.drop_duplicates(subset=['hasil_preprocessing'], inplace=True)
after_dup = len(final_df)

# =========================================================
# SIMPAN HASIL
# =========================================================
output_file = os.path.join(PROGRESS_DIR, 'hasil_preprocessing_dengan_cache.csv')
final_df.to_csv(output_file, index=False)

# Simpan cache terakhir
with open(CACHE_FILE, 'wb') as f:
    pickle.dump(stem_cache, f)

# Hapus checkpoint (sudah selesai)
if os.path.exists(CHECKPOINT_FILE):
    os.remove(CHECKPOINT_FILE)

print(f"\n✓ Hasil disimpan: {output_file}")
print(f"✓ Cache disimpan: {CACHE_FILE} ({len(stem_cache)} kata)")

# =========================================================
# STATISTIK
# =========================================================
print("\n" + "="*60)
print("📊 ANALISIS HASIL PREPROCESSING")
print("="*60)

print(f"\n📍 DATA AWAL:")
print(f"   Total tweet      : {len(df):,}")

print(f"\n📍 DATA KOSONG:")
print(f"   Tweet kosong     : {before_empty - after_empty:,}")

print(f"\n📍 DATA DUPLIKAT:")
print(f"   Tweet duplikat   : {before_dup - after_dup:,}")

print(f"\n📍 DATA FINAL:")
print(f"   Total unique     : {len(final_df):,}")

print(f"\n📍 CACHE:")
print(f"   Kata dalam cache : {len(stem_cache):,}")

print("\n" + "="*60)
print("✅ PREPROCESSING SELESAI!")
print("="*60)

print(f"\n📁 Files saved in: {PROGRESS_DIR}")
print("  ├── hasil_preprocessing_dengan_cache.csv")
print(f"  ├── stemming_cache.pkl ({len(stem_cache):,} kata)")
print("  └── checkpoint_data.csv (sudah dihapus)")
```

```python
# =========================================================
# WORD CLOUD UNTUK TEKS RAW DAN CLEAN (BERSAMBUNGAN)
# =========================================================

import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud, STOPWORDS
import numpy as np
from PIL import Image
import os

# =========================================================
# KONFIGURASI
# =========================================================
PROGRESS_DIR = '/content/drive/MyDrive/Penelitian_@/progress_testing'
OUTPUT_DIR = '/content/drive/MyDrive/Penelitian_@/progress_testing'
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Load hasil preprocessing
file_path = os.path.join(PROGRESS_DIR, 'hasil_preprocessing_dengan_cache.csv')
df = pd.read_csv(file_path)

print("="*60)
print("📊 MEMUAT DATA UNTUK WORD CLOUD")
print("="*60)
print(f"✓ Data loaded: {len(df):,} tweets")

# =========================================================
# FUNGSI UNTUK MEMBUAT WORD CLOUD DENGAN WARNA
# =========================================================
def create_colored_wordcloud(text_series, title, colormap='viridis', max_words=100):
    """
    Membuat word cloud dengan warna

    Parameters:
    - text_series: Series pandas berisi teks
    - title: Judul untuk word cloud
    - colormap: Pilihan warna ('viridis', 'plasma', 'inferno', 'magma', 'coolwarm', 'rainbow')
    - max_words: Jumlah kata maksimal
    """
    # Gabungkan semua teks
    all_text = ' '.join(text_series.astype(str))

    # Tambahkan stopwords tambahan
    additional_stopwords = {'amp', 'rt', 'user', 'co', 'http', 'https', 'tco',
                           'yg', 'tdk', 'ga', 'nya', 'sih', 'dah', 'udah', 'aja'}

    stopwords_set = set(STOPWORDS).union(additional_stopwords)

    # Buat word cloud dengan custom colors
    wordcloud = WordCloud(
        width=1200,
        height=600,
        background_color='white',
        colormap=colormap,
        max_words=max_words,
        stopwords=stopwords_set,
        random_state=42,
        collocations=False,
        normalize_plurals=False,
        min_font_size=10,
        max_font_size=150,
        relative_scaling=0.5,
        prefer_horizontal=0.7
    ).generate(all_text)

    # Plot
    plt.figure(figsize=(14, 8))
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.axis('off')
    plt.title(title, fontsize=20, fontweight='bold', pad=20, fontfamily='sans-serif')
    plt.tight_layout()

    return wordcloud

# =========================================================
# FUNGSI UNTUK WORD CLOUD BERSAMBUNG (SIDE BY SIDE)
# =========================================================
def create_side_by_side_wordcloud(df_raw, df_clean, output_path):
    """
    Membuat dua word cloud bersebelahan
    """
    # Siapkan teks untuk raw
    # Asumsikan kolom raw_text ada di df
    if 'raw_text' in df.columns:
        raw_texts = df['raw_text'].astype(str)
    else:
        # Alternatif: gabungkan dari kolom 'text' asli
        raw_texts = df['hasil_preprocessing'].astype(str)  # fallback

    clean_texts = df['hasil_preprocessing'].astype(str)

    # Buat figure dengan 2 subplot
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 10))

    # Stopwords tambahan
    additional_stopwords = {'amp', 'rt', 'user', 'co', 'http', 'https', 'tco',
                           'yg', 'tdk', 'ga', 'nya', 'sih', 'dah', 'udah', 'aja'}
    stopwords_set = set(STOPWORDS).union(additional_stopwords)

    # Word Cloud 1: RAW TEXT
    all_text_raw = ' '.join(raw_texts)
    wordcloud_raw = WordCloud(
        width=600, height=600,
        background_color='white',
        colormap='Reds',  # Warna merah untuk raw
        max_words=100,
        stopwords=stopwords_set,
        random_state=42,
        collocations=False,
        normalize_plurals=False,
        min_font_size=10,
        max_font_size=120,
        relative_scaling=0.5
    ).generate(all_text_raw)

    # Word Cloud 2: CLEAN TEXT
    all_text_clean = ' '.join(clean_texts)
    wordcloud_clean = WordCloud(
        width=600, height=600,
        background_color='white',
        colormap='Greens',  # Warna hijau untuk clean
        max_words=100,
        stopwords=stopwords_set,
        random_state=42,
        collocations=False,
        normalize_plurals=False,
        min_font_size=10,
        max_font_size=120,
        relative_scaling=0.5
    ).generate(all_text_clean)

    # Plot di subplot kiri (RAW)
    ax1.imshow(wordcloud_raw, interpolation='bilinear')
    ax1.axis('off')
    ax1.set_title('📝 TEKS MENTAH (RAW)\nSebelum Preprocessing',
                  fontsize=18, fontweight='bold', pad=15, color='#CC0000')

    # Plot di subplot kanan (CLEAN)
    ax2.imshow(wordcloud_clean, interpolation='bilinear')
    ax2.axis('off')
    ax2.set_title('✨ TEKS BERSIH (CLEAN)\nSetelah Preprocessing',
                  fontsize=18, fontweight='bold', pad=15, color='#008800')

    # Tambahkan keterangan di bagian bawah
    fig.text(0.5, 0.02,
             f'Berdasarkan {len(df):,} tweet | Total kata unik: {len(set(all_text_clean.split())):,}',
             ha='center', fontsize=12, style='italic')

    plt.suptitle('PERBANDINGAN WORD CLOUD: RAW vs CLEAN',
                 fontsize=22, fontweight='bold', y=0.98)
    plt.tight_layout()

    # Simpan gambar
    plt.savefig(output_path, dpi=300, bbox_inches='tight', facecolor='white')
    print(f"✓ Gambar tersimpan: {output_path}")

    plt.show()

    return wordcloud_raw, wordcloud_clean

# =========================================================
# BEBERAPA VARIASI WORD CLOUD (OPSIONAL)
# =========================================================

def create_multi_wordcloud_variations(df, output_dir):
    """
    Membuat beberapa variasi word cloud untuk raw dan clean
    """
    clean_texts = df['hasil_preprocessing'].astype(str)

    if 'raw_text' in df.columns:
        raw_texts = df['raw_text'].astype(str)
    else:
        raw_texts = clean_texts

    # Stopwords tambahan
    additional_stopwords = {'amp', 'rt', 'user', 'co', 'http', 'https', 'tco',
                           'yg', 'tdk', 'ga', 'nya', 'sih', 'dah', 'udah', 'aja'}
    stopwords_set = set(STOPWORDS).union(additional_stopwords)

    # Variasi 1: Top kata frekuensi tinggi
    print("\n📊 Membuat variasi word cloud...")

    # Untuk RAW
    fig, axes = plt.subplots(2, 2, figsize=(16, 14))
    colormaps = ['plasma', 'viridis', 'coolwarm', 'rainbow']
    titles = ['Plasma (Merah-Kuning)', 'Viridis (Biru-Hijau)',
              'Coolwarm (Biru-Merah)', 'Rainbow (Pelangi)']

    all_text_raw = ' '.join(raw_texts)

    for idx, (ax, cmap, title) in enumerate(zip(axes[0], colormaps, titles)):
        wc = WordCloud(width=500, height=400, background_color='white',
                      colormap=cmap, max_words=80, stopwords=stopwords_set,
                      random_state=42).generate(all_text_raw)
        ax.imshow(wc, interpolation='bilinear')
        ax.axis('off')
        ax.set_title(f'RAW - {title}', fontsize=12, fontweight='bold')

    # Untuk CLEAN
    all_text_clean = ' '.join(clean_texts)

    for idx, (ax, cmap, title) in enumerate(zip(axes[1], colormaps, titles)):
        wc = WordCloud(width=500, height=400, background_color='white',
                      colormap=cmap, max_words=80, stopwords=stopwords_set,
                      random_state=42).generate(all_text_clean)
        ax.imshow(wc, interpolation='bilinear')
        ax.axis('off')
        ax.set_title(f'CLEAN - {title}', fontsize=12, fontweight='bold')

    plt.suptitle('VARIASI WARNA WORD CLOUD (RAW vs CLEAN)',
                 fontsize=18, fontweight='bold', y=0.98)
    plt.tight_layout()

    variasi_path = os.path.join(output_dir, 'wordcloud_variasi_warna.png')
    plt.savefig(variasi_path, dpi=300, bbox_inches='tight')
    print(f"✓ Variasi warna tersimpan: {variasi_path}")
    plt.show()

# =========================================================
# EKSEKUSI UTAMA
# =========================================================

print("\n" + "="*60)
print("🎨 MEMBUAT WORD CLOUD")
print("="*60)

# 1. Buat word cloud bersebelahan (RAW vs CLEAN)
output_sideby = os.path.join(OUTPUT_DIR, 'wordcloud_raw_vs_clean_bersebelahan.png')
wordcloud_raw, wordcloud_clean = create_side_by_side_wordcloud(df, df, output_sideby)

# 2. (Opsional) Buat variasi warna untuk raw dan clean terpisah
create_multi_wordcloud_variations(df, OUTPUT_DIR)

# 3. Buat individual word cloud untuk masing-masing
print("\n📸 Membuat word cloud individual...")

# Word Cloud RAW individual (dengan layout besar)
if 'raw_text' in df.columns:
    raw_texts = df['raw_text'].astype(str)
    all_raw = ' '.join(raw_texts)

    additional_stopwords = {'amp', 'rt', 'user', 'co', 'http', 'https', 'tco'}
    stopwords_set = set(STOPWORDS).union(additional_stopwords)

    wc_raw = WordCloud(width=1200, height=600, background_color='white',
                      colormap='OrRd', max_words=120, stopwords=stopwords_set,
                      random_state=42).generate(all_raw)

    plt.figure(figsize=(16, 8))
    plt.imshow(wc_raw, interpolation='bilinear')
    plt.axis('off')
    plt.title('WORD CLOUD - TEKS MENTAH (RAW)', fontsize=20, fontweight='bold', pad=20)
    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_DIR, 'wordcloud_raw_only.png'), dpi=300, bbox_inches='tight')
    print("✓ Word Cloud RAW (individual) tersimpan")
    plt.show()

# Word Cloud CLEAN individual
clean_texts = df['hasil_preprocessing'].astype(str)
all_clean = ' '.join(clean_texts)

wc_clean = WordCloud(width=1200, height=600, background_color='white',
                    colormap='GnBu', max_words=120, stopwords=stopwords_set,
                    random_state=42).generate(all_clean)

plt.figure(figsize=(16, 8))
plt.imshow(wc_clean, interpolation='bilinear')
plt.axis('off')
plt.title('WORD CLOUD - TEKS BERSIH (CLEAN)', fontsize=20, fontweight='bold', pad=20)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, 'wordcloud_clean_only.png'), dpi=300, bbox_inches='tight')
print("✓ Word Cloud CLEAN (individual) tersimpan")
plt.show()

# =========================================================
# STATISTIK DAN INFORMASI TAMBAHAN
# =========================================================
print("\n" + "="*60)
print("📊 STATISTIK WORD CLOUD")
print("="*60)

# Hitung frekuensi kata dari teks clean
from collections import Counter

all_clean_text = ' '.join(clean_texts)
words = all_clean_text.split()
word_freq = Counter(words)

print(f"\n📍 TOP 10 KATA PALING SERING MUNCUL (CLEAN):")
print("-" * 40)
for i, (word, freq) in enumerate(word_freq.most_common(10), 1):
    print(f"{i:2d}. '{word:<15}' : {freq:>8,} kali")

print(f"\n📍 TOTAL STATISTIK:")
print(f"   Total kata unik (clean)  : {len(word_freq):,}")
print(f"   Total kata keseluruhan   : {len(words):,}")
print(f"   Rata-rata kata per tweet : {len(words)/len(df):.2f}")

print("\n" + "="*60)
print("✅ WORD CLOUD SELESAI!")
print("="*60)
print(f"\n📁 Semua file tersimpan di: {OUTPUT_DIR}")
print("   ├── wordcloud_raw_vs_clean_bersebelahan.png  (2 gambar bersebelahan)")
print("   ├── wordcloud_raw_only.png")
print("   ├── wordcloud_clean_only.png")
print("   └── wordcloud_variasi_warna.png")
```
